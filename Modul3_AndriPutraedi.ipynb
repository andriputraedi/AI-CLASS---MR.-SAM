**10. `hitung_kembalian(total, bayar)`**

````python
def hitung_kembalian(total, bayar):
    if bayar < total:
        return "Uang kurang"
    return bayar - total

print(hitung_kembalian(45000, 50000))   # 5000
print(hitung_kembalian(45000, 40000))   # Uang kurang
````

**11. `diskon(total, persen=10)` dengan nilai bawaan 10%**

````python
def diskon(total, persen=10):
    potongan = total * persen / 100
    return total - potongan

print(diskon(200000))      # 180000.0 (pakai 10% bawaan)
print(diskon(200000, 25))  # 150000.0
````

**12. Tantangan — `statistik(angka)`**

````python
def statistik(angka):
    """Mengembalikan dictionary berisi total, rata-rata, tertinggi, dan terendah."""
    if len(angka) == 0:          # penjagaan kasus khusus: list kosong
        return {"total": 0, "rata_rata": 0, "tertinggi": None, "terendah": None}
    total = 0
    tertinggi = angka[0]
    terendah = angka[0]
    for a in angka:
        total = total + a
        if a > tertinggi:
            tertinggi = a
        if a < terendah:
            terendah = a
    return {
        "total": total,
        "rata_rata": total / len(angka),
        "tertinggi": tertinggi,
        "terendah": terendah,
    }

print(statistik([80, 95, 70, 88, 55]))
````

# Tugas B — Pencatat Pengeluaran Bulanan (dikembangkan)
Fitur tambahan yang dikerjakan (semua 6):
1. Kategori dirapikan otomatis `.lower().strip()`
2. `pengeluaran_terbesar(catatan)`
3. Rata-rata pengeluaran per transaksi
4. Anggaran per kategori + peringatan
5. Mode `while` sampai pengguna mengetik "selesai"
6. Bar persentase sederhana per kategori

### Function-function inti

````python
ANGGARAN = 2000000
ANGGARAN_KATEGORI = {"makan": 800000, "transport": 400000, "belanja": 500000}

def tambah_pengeluaran(catatan, nama, kategori, jumlah):
    kategori = kategori.lower().strip()          # Fitur 1: rapikan kategori
    catatan.append({"nama": nama.strip(), "kategori": kategori, "jumlah": jumlah})
    return catatan

def hitung_total(catatan):
    total = 0
    for item in catatan:
        total = total + item["jumlah"]
    return total

def ringkasan_kategori(catatan):
    ringkasan = {}
    for item in catatan:
        k = item["kategori"]
        ringkasan[k] = ringkasan.get(k, 0) + item["jumlah"]
    return ringkasan

def pengeluaran_terbesar(catatan):              # Fitur 2
    """Mengembalikan satu pengeluaran dengan jumlah paling besar (None bila kosong)."""
    terbesar = None
    for item in catatan:
        if terbesar is None or item["jumlah"] > terbesar["jumlah"]:
            terbesar = item
    return terbesar

def rata_rata_pengeluaran(catatan):             # Fitur 3
    """Rata-rata per transaksi; 0 bila belum ada data."""
    if len(catatan) == 0:
        return 0
    return hitung_total(catatan) / len(catatan)

def cek_anggaran_kategori(ringkasan, batas):    # Fitur 4
    """Mengembalikan list pesan peringatan untuk kategori yang melewati batas."""
    peringatan = []
    for kategori, jumlah in ringkasan.items():
        if kategori in batas and jumlah > batas[kategori]:
            lebih = jumlah - batas[kategori]
            peringatan.append(f"Kategori '{kategori}' melebihi anggaran Rp {batas[kategori]:,} sebesar Rp {lebih:,}")
    return peringatan

def bar_persen(persen, lebar=20):               # Fitur 6
    """Bar sederhana: 40% -> ████████ (lebar maksimum 20 blok)."""
    return "█" * round(persen / 100 * lebar)

def cetak_laporan(catatan):
    total = hitung_total(catatan)
    if total == 0:
        print("Belum ada pengeluaran tercatat.")
        return
    ringkasan = ringkasan_kategori(catatan)
    print("\n========== LAPORAN PENGELUARAN ==========")
    for kategori, jumlah in ringkasan.items():
        persen = jumlah / total * 100
        print(f"{kategori:<10} {bar_persen(persen):<20} {persen:5.1f}%  Rp {jumlah:,}")
    print("=========================================")
    print(f"TOTAL            Rp {total:,}")
    print(f"ANGGARAN         Rp {ANGGARAN:,}")
    print(f"RATA-RATA/TRANS. Rp {rata_rata_pengeluaran(catatan):,.0f}")
    besar = pengeluaran_terbesar(catatan)
    print(f"TERBESAR         {besar['nama']} (Rp {besar['jumlah']:,}, {besar['kategori']})")

    for pesan in cek_anggaran_kategori(ringkasan, ANGGARAN_KATEGORI):
        print("PERINGATAN:", pesan)

    if total > ANGGARAN:
        print(f"\nMELEBIHI ANGGARAN sebesar Rp {total - ANGGARAN:,}")
    elif total > ANGGARAN * 0.8:
        print("\nHati-hati, sudah lebih dari 80% anggaran terpakai.")
    else:
        print(f"\nAman. Sisa anggaran Rp {ANGGARAN - total:,}")
````

### Demo dengan data contoh (perhatikan `"Makan"` dan `"makan "` digabung jadi satu)

````python
catatan = []
tambah_pengeluaran(catatan, "Makan siang",    "Makan",    350000)
tambah_pengeluaran(catatan, "Makan malam",    "makan ",   300000)
tambah_pengeluaran(catatan, "Sarapan",        " MAKAN",   250000)
tambah_pengeluaran(catatan, "Bensin",         "Transport", 200000)
tambah_pengeluaran(catatan, "Sepatu",         "belanja",  450000)
tambah_pengeluaran(catatan, "Pulsa",          "lainnya",  100000)
cetak_laporan(catatan)
````

### Mode interaktif (`while`) — Fitur 5
Ubah `JALANKAN_INTERAKTIF = True` lalu jalankan cell ini untuk mencatat sendiri. Ketik `selesai` pada kolom *Untuk apa* untuk berhenti.

````python
JALANKAN_INTERAKTIF = False

if JALANKAN_INTERAKTIF:
    catatan_baru = []
    while True:
        nama = input("\nUntuk apa (ketik 'selesai' untuk berhenti): ")
        if nama.lower().strip() == "selesai":
            break
        kategori = input("Kategori (makan/transport/belanja/lainnya): ")
        jumlah = int(input("Jumlah (Rp): "))
        tambah_pengeluaran(catatan_baru, nama, kategori, jumlah)
    cetak_laporan(catatan_baru)
````

# Tugas C — Latihan Membaca Error
Kode yang rusak ditampilkan sebagai teks (supaya notebook tetap bisa di-*Run all*), lalu diikuti perbaikannya.
> Catatan: untuk memenuhi instruksi "salin pesan error dari Colab", tempel kode rusak ke sebuah cell dan jalankan sendiri; pesan di bawah adalah pesan yang akan muncul.

### Error 1
```python
nilai = [80, 90, 70]
for n in nilai
    print(n)
```
**Pesan error:** `SyntaxError: expected ':'` (di Python versi lama: `SyntaxError: invalid syntax`)

**Artinya (bahasa saya):** Python menemukan baris `for` yang tidak diakhiri titik dua `:`. Titik dua wajib ada di akhir `if`, `for`, `while`, dan `def` untuk memberi tahu Python bahwa blok kode dimulai.

**Perbaikan:** tambahkan `:` di akhir baris `for`.

````python
nilai = [80, 90, 70]
for n in nilai:
    print(n)
````

### Error 2
```python
def hitung(a, b):
return a + b
```
**Pesan error:** `IndentationError: expected an indented block after function definition on line 1`

**Artinya:** Baris setelah `def` harus menjorok (indentasi 4 spasi). Di sini `return` ditulis rata kiri, jadi Python tidak tahu mana isi function-nya.

**Perbaikan:** beri 4 spasi sebelum `return`.

````python
def hitung(a, b):
    return a + b

print(hitung(2, 3))   # 5
````

### Error 3
```python
harga = input("Harga: ")
print(harga + 5000)
```
**Pesan error:** `TypeError: can only concatenate str (not "int") to str`

**Artinya:** `input()` selalu menghasilkan teks (*string*), sedangkan `5000` adalah angka (*integer*). Python tidak bisa menjumlahkan teks dengan angka.

**Perbaikan:** bungkus hasil `input()` dengan `int()`.

````python
harga = int(input("Harga: "))
print(harga + 5000)
````

# Tugas D — Chatbot FAQ v0 (Warung Kopi Senja)
Usaha yang dipilih: **warung kopi** (contoh usaha fiktif berbasis usaha kecil umum). Ganti isi `faq` dengan data usaha nyata yang kamu kenal bila diperlukan.

### 1. Basis pengetahuan (7 entri)

````python
faq = [
    {"kata_kunci": ["jam", "buka", "tutup", "operasional"],
     "jawaban": "Kami buka setiap hari pukul 08.00-22.00 WIB."},
    {"kata_kunci": ["menu", "makanan", "minuman", "kopi", "jual"],
     "jawaban": "Menu andalan kami: kopi susu, es teh, roti bakar, dan mie goreng."},
    {"kata_kunci": ["harga", "berapa", "biaya"],
     "jawaban": "Kopi mulai Rp 12.000, makanan mulai Rp 15.000."},
    {"kata_kunci": ["bayar", "pembayaran", "transfer", "qris"],
     "jawaban": "Pembayaran bisa lewat QRIS, transfer bank, atau tunai."},
    {"kata_kunci": ["alamat", "lokasi", "dimana", "tempat"],
     "jawaban": "Kami di Jl. Contoh No. 10, Jakarta."},
    {"kata_kunci": ["wifi", "internet", "colokan", "charger"],
     "jawaban": "Wi-Fi gratis dan colokan tersedia di semua meja."},
    {"kata_kunci": ["antar", "delivery", "kirim", "ongkir"],
     "jawaban": "Tersedia pesan antar lewat ojek online, gratis ongkir untuk radius 2 km."},
]
````

### 2. Function chatbot (+ sapaan)

````python
SAPAAN = ["halo", "hai", "hallo", "hi"]

def bersihkan_teks(teks):
    """Merapikan input pengguna agar mudah dicocokkan."""
    teks = teks.lower().strip()
    for tanda in ["?", "!", ".", ","]:
        teks = teks.replace(tanda, "")
    return teks

def cari_jawaban(pertanyaan, basis):
    """Mencari entri FAQ paling cocok.

    pertanyaan: teks yang sudah dibersihkan
    basis: list berisi dictionary kata_kunci dan jawaban
    Mengembalikan: (jawaban, skor kecocokan)
    """
    kata_pengguna = pertanyaan.split()
    jawaban_terbaik = ""
    skor_tertinggi = 0
    for entri in basis:
        skor = 0
        for kata in kata_pengguna:
            if kata in entri["kata_kunci"]:
                skor = skor + 1
        if skor > skor_tertinggi:
            skor_tertinggi = skor
            jawaban_terbaik = entri["jawaban"]
    return jawaban_terbaik, skor_tertinggi

def balas(pertanyaan, basis, ambang=1):
    """Menyusun balasan akhir untuk pengguna (termasuk sapaan)."""
    bersih = bersihkan_teks(pertanyaan)
    if bersih in SAPAAN:                       # sapaan tidak dicari di FAQ
        return "Halo! Selamat datang di Warung Kopi Senja. Ada yang bisa kami bantu?"
    jawaban, skor = cari_jawaban(bersih, basis)
    if skor >= ambang:
        return jawaban
    return "Maaf, saya belum punya informasi itu. Silakan hubungi admin di 0812-xxxx."

def statistik_percakapan(pertanyaan_gagal, total):
    """Mengembalikan dictionary: jumlah pertanyaan, jumlah gagal, persentase keberhasilan."""
    gagal = len(pertanyaan_gagal)
    if total == 0:                             # penjagaan pembagian dengan nol
        return {"jumlah_pertanyaan": 0, "jumlah_gagal": 0, "persentase_berhasil": 0.0}
    berhasil = (total - gagal) / total * 100
    return {"jumlah_pertanyaan": total, "jumlah_gagal": gagal,
            "persentase_berhasil": round(berhasil, 1)}
````

### 3. Mode chat interaktif
Ubah flag menjadi `True` untuk ngobrol langsung dengan bot. Ketik `selesai` untuk berhenti.

````python
JALANKAN_CHAT = False

if JALANKAN_CHAT:
    pertanyaan_gagal = []
    total = 0
    print("=== CHATBOT WARUNG KOPI SENJA (ketik 'selesai' untuk berhenti) ===")
    while True:
        tanya = input("\nAnda : ")
        if bersihkan_teks(tanya) == "selesai":
            break
        total = total + 1
        jawab = balas(tanya, faq)
        print("Bot  :", jawab)
        if jawab.startswith("Maaf"):
            pertanyaan_gagal.append(tanya)
    print(statistik_percakapan(pertanyaan_gagal, total))
````

### 4. Uji 10 pertanyaan (7 seharusnya terjawab, 3 seharusnya gagal)

````python
uji = [
    # (pertanyaan, seharusnya terjawab?)
    ("Jam buka jam berapa?",             True),
    ("Ada menu apa saja?",               True),
    ("Berapa harga kopi susu?",          True),
    ("Bisa bayar pakai QRIS?",           True),
    ("Alamat tokonya dimana?",           True),
    ("Ada wifi gak?",                    True),
    ("Bisa delivery ke rumah?",          True),
    ("Kapan tokonya mulai beroperasi?",  False),   # sinonim, tidak ada di kata kunci
    ("Apakah ada diskon mahasiswa?",     False),   # info tidak ada di FAQ
    ("Kapan oprasional hari libur?",    False),   # salah ketik "oprasional"
]

pertanyaan_gagal = []
cocok_ekspektasi = 0
print(f"{'No':<3} {'Pertanyaan':<35} {'Harapan':<10} {'Hasil':<10}")
print("-" * 62)
for i, (tanya, harus_terjawab) in enumerate(uji, start=1):
    jawab = balas(tanya, faq)
    gagal = jawab.startswith("Maaf")
    if gagal:
        pertanyaan_gagal.append(tanya)
    sesuai = (not gagal) == harus_terjawab
    if sesuai:
        cocok_ekspektasi = cocok_ekspektasi + 1
    print(f"{i:<3} {tanya:<35} {'jawab' if harus_terjawab else 'gagal':<10} "
          f"{'gagal' if gagal else 'jawab':<10} {'OK' if sesuai else 'BEDA'}")

print()
print("Pertanyaan yang gagal dijawab:")
for p in pertanyaan_gagal:
    print("-", p)
print()
print(statistik_percakapan(pertanyaan_gagal, len(uji)))
print(f"Hasil sesuai ekspektasi: {cocok_ekspektasi} dari {len(uji)}")
print(balas("Halo", faq))   # tes sapaan
````
